```python
#Purnomo Jeremy Bambang_202431026
cat_col = ["Sex", "Cabin", "Embarked"]

for col in cat_col:
    df[col] = df[col].astype('category')

df.dtypes
```

```python
#Purnomo Jeremy Bambang_202431026
for col in cat_col:
    print(col, ":", df[col].unique())
```

## Detection and Treatment of Missing Values

```python
df.isna().sum() #Purnomo Jeremy Bambang_202431026
```

```python
#Purnomo Jeremy Bambang_202431026
for col in df.columns:
    if pd.api.types.is_numeric_dtype(df[col]):
        df[col] = df[col].fillna(df[col].mean()) #angka diisi mean
    else:
        df[col] = df[col].fillna(df[col].mode()[0]) #kategorik diisi modus

df.isna().sum()
```

```python
print("Jumlah data duplikat: ", df.duplicated().sum()) #Purnomo Jeremy Bambang_202431026
```

```python
#Purnomo Jeremy Bambang_202431026
df = df.drop_duplicates()
df.info()
```

```python
#Purnomo Jeremy Bambang_202431026
sns.boxplot(x="Age", data = df)
plt.title("Distribusi Umur")
plt.show()
```

```python
#Purnomo Jeremy Bambang_202431026
Q1 = df['Age'].quantile(1/4) #quartile 1 = 25% data bawah
Q3 = df['Age'].quantile(3/4) #quartile 3 = 75% data

IQR = Q3-Q1

lower_bound = Q1 - 3/2 * IQR
upper_bound = Q3 + 3/2 * IQR

outliers = df[(df["Age"] < lower_bound) | (df["Age"] > upper_bound)]
print("Jumlah outlier umur: ", len(outliers)) 
```

## Correlation of Variable
