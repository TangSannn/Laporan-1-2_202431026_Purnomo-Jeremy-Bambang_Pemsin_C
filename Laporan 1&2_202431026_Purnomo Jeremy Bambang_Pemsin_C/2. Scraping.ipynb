```python
import requests
import csv
from bs4 import BeautifulSoup
import time
# Purnomo Jeremy Bambang_202431026
```

```python
headers = {'User-Agent': 'PraktikumPemsin Purnomo Jeremy Bambang_202431026'}
response = requests.get('https://id.wikipedia.org/wiki/Daftar_kabupaten_di_Indonesia', headers=headers)
print(response.status_code)
print(response.text[:500])
# Purnomo Jeremy Bambang_202431026
```

```python
soup = BeautifulSoup(response.text, "html.parser")
tabel = soup.find("table", class_="wikitable")
blocks = tabel.find_all("tr")[2:] #2 baris pertama adalah header
print(f"kabupaten yang ditemukan pada website ini sejumlah: {len(blocks)}")
# Purnomo Jeremy Bambang_202431026
```

```python
result = [] #Purnomo Jeremy Bambang_202431026
for block in blocks:
    kolom = block.find_all(["th", "td"])
    if len(kolom) != 8: #baris kabupaten harus punya 8 sel
        continue
    nama = kolom[1].get_text(strip=True)
    provinsi = kolom[3].get_text(strip=True)
    kecamatan = kolom[5].get_text(strip=True)
    kelurahan = kolom[6].get_text(strip=True)
    desa = kolom[7].get_text(strip=True)

    result.append({"nama": nama, "provinsi": provinsi, "kecamatan": kecamatan,
                   "kelurahan": kelurahan, "desa": desa, "jenis": "Kabupaten"})

print(len(result), "data masuk,", len(blocks) - len(result), "baris dilewati")
```

```python
time.sleep(2)
response = requests.get('https://id.wikipedia.org/wiki/Daftar_kota_di_Indonesia', headers=headers)
print(response.status_code)

soup = BeautifulSoup(response.text, "html.parser")
tabel = soup.find("table", class_="wikitable")
blocks = tabel.find_all("tr")[2:]
print(f"kota yang ditemukan pada website ini sejumlah: {len(blocks)}")

jumlah_kab = len(result)
for block in blocks:
    kolom = block.find_all(["th", "td"])
    if len(kolom) != 7: #baris kota tidak punya kolom ibu kota, jadi 7 sel
        continue
    nama = kolom[1].get_text(strip=True)
    provinsi = kolom[3].get_text(strip=True)
    kecamatan = kolom[4].get_text(strip=True)
    kelurahan = kolom[5].get_text(strip=True)
    desa = kolom[6].get_text(strip=True)

    result.append({"nama": nama, "provinsi": provinsi, "kecamatan": kecamatan,
                   "kelurahan": kelurahan, "desa": desa, "jenis": "Kota"})

print(len(result) - jumlah_kab, "data kota masuk,", len(blocks) - (len(result) - jumlah_kab), "baris dilewati")
print("Total data:", len(result))
# Purnomo Jeremy Bambang_202431026
```

```python
for item in result:
    print(f"{item['jenis']}: {item['nama']} - provinsi: {item['provinsi']} - kecamatan: {item['kecamatan']} - kelurahan: {item['kelurahan']} - desa: {item['desa']}")

# Purnomo Jeremy Bambang_202431026
```

```python
with open("wilayah_indonesia.csv", "w", newline="", encoding="utf-8") as csvfile:
    fieldnames = ["nama", "provinsi", "kecamatan", "kelurahan", "desa", "jenis"]
    writer = csv.DictWriter(csvfile, fieldnames=fieldnames)

    writer.writeheader()
    for item in result:
        writer.writerow(item)

# Purnomo Jeremy Bambang_202431026
```
